```python
import os
os.chdir("D:\Documents\Projects\image-captioning")
print("Working directory:", os.getcwd())

import yaml
import torch
import matplotlib.pyplot as plt
from PIL import Image

from src.vocabulary import Vocabulary
from src.transform import get_eval_transform
from src.model import CaptionModel
```

```python
with open("config.yaml", "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

vocab = Vocabulary.load(config["paths"]["vocab"])
print("Vocab size:", len(vocab))

model = CaptionModel(
    embed_size=config["model"]["embed_size"],
    hidden_size=config["model"]["hidden_size"],
    vocab_size=len(vocab),
    num_layers=config["model"]["num_layers"],
).to(device)

checkpoint = torch.load(config["paths"]["checkpoint"], map_location=device, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

print(f"Đã load checkpoint, epoch={checkpoint.get('epoch')}, val_loss={checkpoint.get('val_loss'):.4f}")

transform = get_eval_transform(config["data"]["image_size"])
```

```python
def predict_from_path(image_path, method="greedy", beam_size=3, show=True):
    """
    Nhận vào đường dẫn ảnh bất kỳ (local path), trả về caption dự đoán.
    """
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"Không tìm thấy ảnh tại: {image_path}")

    raw_image = Image.open(image_path).convert("RGB")
    image_tensor = transform(raw_image).unsqueeze(0).to(device)  # (1, 3, H, W)

    plt.imshow(raw_image)
    plt.axis('off')
    plt.title(image_path)

    with torch.no_grad():
        caption = model.generate_caption(
            image_tensor,
            vocab,
            max_len=config["data"]["max_len"],
            method=method,
            beam_size=beam_size,
        )[0]

    print(f"Ảnh: {os.path.basename(image_path)}")
    plt.show()
    print(f"Caption ({method}): {caption}")
    return caption
```

```python
image_path = r"C:\Users\Admin\Pictures\Saved Pictures\DSC_0580 copy.jpg"

predict_from_path(image_path, method="greedy")
predict_from_path(image_path, method="beam")
```
