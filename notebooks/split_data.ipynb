```python
# Move the root directory to the system path 
import sys
from pathlib import Path

root_dir = Path().resolve().parent
if str(root_dir) not in sys.path:
    sys.path.append(str(root_dir))
```

```python
import json
from collections import defaultdict
from src.utils import save_splits
```

```python
# Karpathy split
train = defaultdict(list)
val =  defaultdict(list)
test = defaultdict(list)

with open('../data/raw/dataset_flickr30k.json', 'r') as f:
    data = json.load(f)
    for item in data["images"]:
        split = item["split"]
        if split == "train":
            for sentence in item["sentences"]:
                train[item["filename"]].append(sentence["raw"])
        elif split == "val":
            for sentence in item["sentences"]:
                val[item["filename"]].append(sentence["raw"])
        elif split == "test":
            for sentence in item["sentences"]:
                test[item["filename"]].append(sentence["raw"])
```

```python
save_splits(train, val, test)
```
